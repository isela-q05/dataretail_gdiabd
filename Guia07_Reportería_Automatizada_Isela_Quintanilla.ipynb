```python
# IMPORTAR LIBRERÍAS
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt              # Gráficos estáticos (estable)
from pathlib import Path
from datetime import datetime
from jinja2 import Template
import base64
import json
import warnings
warnings.filterwarnings('ignore')  # Ignorar warnings menores

# CONFIGURAR MATPLOTLIB
plt.rcParams['figure.figsize'] = (12, 5)
plt.rcParams['font.size'] = 11
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False

print(" Librerías importadas")
print(f"   - pandas: {pd.__version__}")
print(f"   - matplotlib: {plt.matplotlib.__version__}")

# CARGAR DATASETS
df = pd.read_csv(BASE / 'datos' / 'procesados' / 'dataset_limpio_guia2_Isela_Quintanilla.csv')
df['fecha'] = pd.to_datetime(df['fecha'])

df_sucursales = pd.read_csv(BASE / 'datos' / 'geo' / 'sucursales_dataretail_guia06.csv')
df_demanda = pd.read_csv(BASE / 'datos' / 'geo' / 'demanda_dataretail_guia06.csv')

print(f"\nDatasets cargados:")
print(f"   - Ventas: {df.shape[0]} filas × {df.shape[1]} columnas")
print(f"   - Sucursales: {df_sucursales.shape[0]} filas")
print(f"   - Demanda: {df_demanda.shape[0]} filas")

print(f"\nPrimeras 3 filas del dataset principal:")
print(df.head(3).to_string())
```

```python
# CALCULAR KPIs
kpis = {
    'total_ventas':       df['total_ventas'].sum(),
    'numero_ventas':      df['id_venta'].nunique(),
    'unidades_vendidas':  df['cantidad'].sum(),
    'utilidad_estimada':  df['utilidad_estimada'].sum(),
    'margen_pct':         (df['utilidad_estimada'].sum() / df['total_ventas'].sum()) * 100,
    'ticket_promedio':    df['total_ventas'].sum() / df['id_venta'].nunique(),
    'fecha_inicio':       df['fecha'].min().strftime('%d/%m/%Y'),
    'fecha_fin':          df['fecha'].max().strftime('%d/%m/%Y'),
    'periodo':            f"{df['fecha'].min().strftime('%d/%m/%Y')} - {df['fecha'].max().strftime('%d/%m/%Y')}"
}

print("KPIs DEL PERIODO")
print("=" * 50)
print(f"Total Ventas:        ${kpis['total_ventas']:>10,.2f}")
print(f"Número de Ventas:    {kpis['numero_ventas']:>10}")
print(f"Unidades Vendidas:   {kpis['unidades_vendidas']:>10}")
print(f"Utilidad Estimada:   ${kpis['utilidad_estimada']:>10,.2f}")
print(f"Margen %:            {kpis['margen_pct']:>9.1f}%")
print(f"Ticket Promedio:     ${kpis['ticket_promedio']:>10,.2f}")
print(f"Periodo:             {kpis['periodo']}")
print("=" * 50)
```

```python
#ANÁLISIS POR CATEGORÍA
df_categoria = df.groupby('categoria').agg(
    ventas=('total_ventas', 'sum'),
    utilidad=('utilidad_estimada', 'sum'),
    operaciones=('id_venta', 'count')
).reset_index()
df_categoria['margen'] = (df_categoria['utilidad'] / df_categoria['ventas']) * 100
df_categoria = df_categoria.sort_values('ventas', ascending=False)

print("ANÁLISIS POR CATEGORÍA:")
print(df_categoria.to_string(index=False))

#ANÁLISIS POR CANAL
df_canal = df.groupby('canal').agg(
    ventas=('total_ventas', 'sum'),
    utilidad=('utilidad_estimada', 'sum'),
    operaciones=('id_venta', 'count')
).reset_index()
df_canal['margen'] = (df_canal['utilidad'] / df_canal['ventas']) * 100
df_canal = df_canal.sort_values('ventas', ascending=False)

print("\nANÁLISIS POR CANAL:")
print(df_canal.to_string(index=False))

#ANÁLISIS POR REGIÓN (Guía 6)
df_region = df_sucursales.groupby('region').agg(
    ventas_totales=('ventas_mes', 'sum'),
    variacion_promedio=('variacion_pct', 'mean'),
    sucursales=('municipio', 'count')
).reset_index()
df_region = df_region.sort_values('ventas_totales', ascending=False)

print("\nANÁLISIS POR REGIÓN:")
print(df_region.to_string(index=False))
```

```python
# ===================================================================
# GRÁFICO 1: Tendencia temporal (líneas)
# ===================================================================
fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(df['fecha'], df['total_ventas'],
        marker='o', linewidth=2.5, markersize=10,
        color='#2c7bb6', markerfacecolor='#1a5490')
ax.fill_between(df['fecha'], df['total_ventas'], alpha=0.2, color='#2c7bb6')
ax.set_title('Evolución de Ventas por Fecha', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Fecha', fontsize=11)
ax.set_ylabel('Ventas ($)', fontsize=11)
ax.grid(True, alpha=0.3, linestyle='--')
ax.tick_params(axis='x', rotation=45)

# Añadir etiquetas con valores
for fecha, valor in zip(df['fecha'], df['total_ventas']):
    ax.annotate(f'${valor:.2f}', (fecha, valor),
                textcoords="offset points", xytext=(0, 10),
                ha='center', fontsize=9, fontweight='bold')

plt.tight_layout()
ruta_tendencia = BASE / 'evidencias' / 'reporte' / 'grafico_tendencia.png'
plt.savefig(ruta_tendencia, dpi=150, bbox_inches='tight', facecolor='white')
plt.close()
print(f"------ Gráfico 1: {ruta_tendencia.name}")

# ===================================================================
# GRÁFICO 2: Ventas por categoría (barras)
# ===================================================================
fig, ax = plt.subplots(figsize=(10, 6))
colores = plt.cm.Blues(np.linspace(0.4, 0.9, len(df_categoria)))
barras = ax.bar(df_categoria['categoria'], df_categoria['ventas'],
                color=colores, edgecolor='#1a5490', linewidth=1.5)
ax.set_title('Ventas por Categoría', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Categoría', fontsize=11)
ax.set_ylabel('Ventas ($)', fontsize=11)
ax.grid(True, alpha=0.3, axis='y', linestyle='--')

for barra, valor in zip(barras, df_categoria['ventas']):
    height = barra.get_height()
    ax.text(barra.get_x() + barra.get_width()/2., height,
            f'${valor:.2f}', ha='center', va='bottom',
            fontsize=11, fontweight='bold')

plt.tight_layout()
ruta_categoria = BASE / 'evidencias' / 'reporte' / 'grafico_categorias.png'
plt.savefig(ruta_categoria, dpi=150, bbox_inches='tight', facecolor='white')
plt.close()
print(f"------ Gráfico 2: {ruta_categoria.name}")

# ===================================================================
# GRÁFICO 3: Ventas por canal (barras)
# ===================================================================
fig, ax = plt.subplots(figsize=(10, 6))
colores = ['#e74c3c', '#3498db', '#2ecc71']
barras = ax.bar(df_canal['canal'], df_canal['ventas'],
                color=colores, edgecolor='#2c3e50', linewidth=1.5)
ax.set_title('Ventas por Canal', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Canal', fontsize=11)
ax.set_ylabel('Ventas ($)', fontsize=11)
ax.grid(True, alpha=0.3, axis='y', linestyle='--')

for barra, valor in zip(barras, df_canal['ventas']):
    height = barra.get_height()
    ax.text(barra.get_x() + barra.get_width()/2., height,
            f'${valor:.2f}', ha='center', va='bottom',
            fontsize=11, fontweight='bold')

plt.tight_layout()
ruta_canal = BASE / 'evidencias' / 'reporte' / 'grafico_canal.png'
plt.savefig(ruta_canal, dpi=150, bbox_inches='tight', facecolor='white')
plt.close()
print(f"------ Gráfico 3: {ruta_canal.name}")

# ===================================================================
# GRÁFICO 4: Ventas por región (barras)
# ===================================================================
fig, ax = plt.subplots(figsize=(10, 6))
colores = plt.cm.Greens(np.linspace(0.4, 0.9, len(df_region)))
barras = ax.bar(df_region['region'], df_region['ventas_totales'],
                color=colores, edgecolor='#1a5490', linewidth=1.5)
ax.set_title('Ventas por Región (Análisis Geoespacial)', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Región', fontsize=11)
ax.set_ylabel('Ventas ($)', fontsize=11)
ax.grid(True, alpha=0.3, axis='y', linestyle='--')

for barra, valor in zip(barras, df_region['ventas_totales']):
    height = barra.get_height()
    ax.text(barra.get_x() + barra.get_width()/2., height,
            f'${valor:,.0f}', ha='center', va='bottom',
            fontsize=11, fontweight='bold')

plt.tight_layout()
ruta_region = BASE / 'evidencias' / 'reporte' / 'grafico_region.png'
plt.savefig(ruta_region, dpi=150, bbox_inches='tight', facecolor='white')
plt.close()
print(f"------ Gráfico 4: {ruta_region.name}")

print("\n¡Los 4 gráficos se generaron correctamente SIN reiniciar Colab!")
```

```python
# GENERAR DOCX CON PYTHON-DOCX
from docx import Document
from docx.shared import Inches
from docx.enum.text import WD_ALIGN_PARAGRAPH

doc = Document()

# Título
titulo = doc.add_heading('INFORME GERENCIAL DE VENTAS', 0)
titulo.alignment = WD_ALIGN_PARAGRAPH.CENTER

subtitulo = doc.add_heading('DataRetail AI', level=1)
subtitulo.alignment = WD_ALIGN_PARAGRAPH.CENTER

doc.add_paragraph(f'Periodo: {kpis["periodo"]}')
doc.add_paragraph(f'Fecha de generación: {datetime.now().strftime("%d/%m/%Y %H:%M")}')
doc.add_paragraph(f'Versión: 1.0')

# Resumen ejecutivo
doc.add_heading('1. Resumen Ejecutivo', level=2)
doc.add_paragraph(f'El presente informe analiza el desempeño comercial de DataRetail AI durante el periodo {kpis["periodo"]}.')

# KPIs
doc.add_heading('Indicadores Principales', level=3)
tabla_kpis = doc.add_table(rows=6, cols=2)
tabla_kpis.style = 'Light Grid Accent 1'

datos_kpis = [
    ('Total Ventas', f'${kpis["total_ventas"]:,.2f}'),
    ('Número de Ventas', f'{kpis["numero_ventas"]}'),
    ('Unidades Vendidas', f'{kpis["unidades_vendidas"]}'),
    ('Utilidad Estimada', f'${kpis["utilidad_estimada"]:,.2f}'),
    ('Margen %', f'{kpis["margen_pct"]:.1f}%'),
    ('Ticket Promedio', f'${kpis["ticket_promedio"]:,.2f}')
]

for i, (label, valor) in enumerate(datos_kpis):
    tabla_kpis.rows[i].cells[0].text = label
    tabla_kpis.rows[i].cells[1].text = valor

# Resultados con gráficos
doc.add_heading('2. Resultados Generales', level=2)
doc.add_heading('2.1 Evolución de Ventas', level=3)
doc.add_picture(str(ruta_tendencia), width=Inches(6))

doc.add_heading('2.2 Ventas por Categoría', level=3)
doc.add_picture(str(ruta_categoria), width=Inches(6))

doc.add_heading('2.3 Ventas por Canal', level=3)
doc.add_picture(str(ruta_canal), width=Inches(6))

doc.add_heading('2.4 Análisis Geoespacial', level=3)
doc.add_picture(str(ruta_region), width=Inches(6))

# Hallazgos
doc.add_heading('3. Hallazgos Principales', level=2)
for i, h in enumerate(datos_template['hallazgos'], 1):
    doc.add_heading(f'{i}. {h["titulo"]}', level=3)
    doc.add_paragraph(f'Evidencia: {h["evidencia"]}')
    doc.add_paragraph(f'Interpretación: {h["interpretacion"]}')
    doc.add_paragraph(f'Impacto: {h["impacto"]}')
    doc.add_paragraph(f'Decisión: {h["decision"]}')

# Recomendaciones
doc.add_heading('4. Recomendaciones', level=2)
tabla_rec = doc.add_table(rows=1, cols=5)
tabla_rec.style = 'Light Grid Accent 1'
headers = ['#', 'Acción', 'Prioridad', 'Responsable', 'Plazo']
for i, h in enumerate(headers):
    tabla_rec.rows[0].cells[i].text = h

for i, r in enumerate(datos_template['recomendaciones'], 1):
    row = tabla_rec.add_row()
    row.cells[0].text = str(i)
    row.cells[1].text = r['accion']
    row.cells[2].text = r['prioridad']
    row.cells[3].text = r['responsable']
    row.cells[4].text = r['plazo']

# Limitaciones
doc.add_heading('5. Limitaciones', level=2)
for l in datos_template['limitaciones']:
    doc.add_paragraph(l, style='List Bullet')

# Guardar
ruta_docx = BASE / 'exportaciones' / 'informe_dataretail_v1.0.docx'
doc.save(str(ruta_docx))
print(f"DOCX generado: {ruta_docx.name}")
print(f"📏 Tamaño: {ruta_docx.stat().st_size / 1024:.1f} KB")

files.download(str(ruta_docx))
print("DOCX descargado")
```
